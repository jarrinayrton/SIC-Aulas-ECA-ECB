```python

# Un CONDICIONAL ejecuta código solo si se cumple una condición.
# Definimos los datos iniciales
temperatura = 18
esta_lloviendo = True

print("--- ASISTENTE DE ROPA VIRTUAL ---\n")

# 1. EL IF MÁS BÁSICO (Solo evalúa una cosa)
# Si la condición es cierta (True), ejecuta el código adentro. Si es falsa, lo ignora.
if esta_lloviendo == True:
    print("¡Ojo! Está lloviendo, no olvides llevar tu paraguas.")


# 2. IF y ELSE (Si pasa esto, haz esto... de lo contrario, haz lo otro)
# Evaluamos si hace calor para ir a la playa
if temperatura >= 28:
    print(" Hace mucho calor, es un día perfecto para ir a la playa.")
else:
    print("No hace suficiente calor para la playa, mejor nos quedamos en la ciudad.")


# 3. IF, ELIF y ELSE (Para evaluar múltiples escenarios en orden)
# Dependiendo de los grados exactos, te recomendamos qué abrigo usar:
print("\n--- RECOMENDACIÓN DE ROPA ---")

if temperatura < 10:
    print("Hace mucho frío. ¡Ponte un abrigo grueso, guantes y bufanda!")
elif temperatura <= 20:
    print("El clima está fresco. Una chaqueta ligera o un suéter es suficiente.")
elif temperatura <= 27:
    print(" El clima está muy agradable. Puedes salir en manga corta.")
else:
    print(" Hace muchísimo calor. ¡Busca ropa muy fresca y toma agua!")

```

```python
# BUCLE FOR: Se usa cuando sabes exactamente cuántas veces quieres repetir algo.
# range(inicio, fin) genera números desde el inicio hasta un número antes del fin.

print("--- PRUEBA DE SONIDO (for) ---")
# Repetimos una acción 3 veces (contando 1, 2 y 3)
for intento in range(1, 4):
    print(f"Probando micrófono... Intento número {intento}")
print("Prueba terminada. El equipo funciona.")

print("\n--- CUENTA REGRESIVA (for hacia atrás) ---")
# range(inicio, fin, paso). Empezamos en 5, paramos al llegar a 0, restando 1.
for segundo in range(5, 0, -1):
    print(f"Faltan {segundo} segundos...")
print("¡Tiempo terminado!")

# BUCLE WHILE: Se usa cuando NO sabes cuántas veces vas a repetir algo, 
# pero sabes que debe detenerse cuando se cumpla una condición.

print("\n--- LLENANDO UNA BATERÍA (while) ---")
porcentaje_bateria = 0

# "Mientras" la batería sea menor a 100, seguimos cargando.
while porcentaje_bateria < 100:
    print(f"Batería al {porcentaje_bateria}%. Cargando...")
    porcentaje_bateria += 25  # Aumentamos la carga de 25 en 25

print("Batería al 100%. Dispositivo listo.")


# CONTROL DE BUCLES: break (para detener el bucle por completo)

print("\n--- BUSCANDO LA LLAVE CORRECTA (while con break) ---")
llave_probada = 1

while llave_probada <= 5:
    if llave_probada == 3:
        print(f"Probando llave {llave_probada}: ¡Es la correcta! Puerta abierta.")
        break  # 'break' rompe el ciclo inmediatamente, ya no prueba la 4 ni la 5
    
    print(f"Probando llave {llave_probada}: No funciona.")
    llave_probada += 1  # Pasamos a la siguiente llave
```
