# EJERCICIO 1 · Cuenta bancaria
Construimos una clase `CuentaBancaria` paso a paso, siguiendo la misión de la unidad 21 (diap. 293-300). Una **clase** es el plano; cada **objeto** o **instancia** es una cuenta concreta con sus propios datos.

### 1.1 Definir la clase: constructor, métodos y `__str__`

creamos la clase con su constructor `__init__`, métodos para depositar y retirar, y `__str__` para mostrarla como texto. Incluye una variable de clase, un contador de cuentas y un atributo **privado** para el saldo.



```python
class CuentaBancaria:  
    TASA_BASE = 0.02  # VARIABLE DE CLASE
    total_cuentas = 0  

    def __init__(self, titular, numero, saldo=0):  # CONSTRUCTOR
        self.titular = titular 
        self.numero = numero 
        self.__saldo = saldo  # doble guion bajo: atributo PRIVADO
        CuentaBancaria.total_cuentas += 1 

    def obtener_saldo(self):  
        return self.__saldo  

    def depositar(self, importe): 
        self.__saldo += importe  
        print("Se ha ingresado ${}. El saldo es ${}.".format(importe, self.__saldo))  

    def retirar(self, importe): 
        if importe > self.__saldo:  
            print("Saldo insuficiente: ${} es menos que ${}.".format(self.__saldo, importe)) 
        else:
            self.__saldo -= importe  
            print("Se ha retirado ${}. El saldo es ${}.".format(importe, self.__saldo))  

    def calcular_interes(self): 
        return self.__saldo * CuentaBancaria.TASA_BASE 

    def __str__(self):  
        return "Cuenta {} de {}: ${}".format(self.numero, self.titular, self.__saldo) 
```

### 1.2 Crear instancias y usar los métodos
comprobar que cada instancia tiene sus propios datos pero comparte la misma clase (diap. 316-323 y 330).

```python
cuenta1 = CuentaBancaria("David", "1234-0001")  
cuenta2 = CuentaBancaria("Laura", "1234-0002", 500) 

print(cuenta1)  
cuenta1.depositar(2000) 
cuenta1.retirar(500) 
cuenta1.retirar(5000)  
cuenta2.depositar(100)  
print(cuenta1)  
print(cuenta2)
```

### 1.3 Herencia

creamos `CuentaAhorro`, una clase **hija** que hereda todo de `CuentaBancaria` y agrega una tasa propia y un método para aplicar interés.


```python
class CuentaAhorro(CuentaBancaria): 
    def __init__(self, titular, numero, saldo=0, tasa=0.05):  # constructor propio de la hija
        CuentaBancaria.__init__(self, titular, numero, saldo) 
        self.tasa = tasa  # atributo nuevo

    def aplicar_interes(self):  # método nuevo de la hija
        interes = self.obtener_saldo() * self.tasa  
        self.depositar(interes)  

ahorro = CuentaAhorro("Marta", "1234-0003", 1000)  
ahorro.aplicar_interes()  
ahorro.retirar(100)  
print(CuentaBancaria.total_cuentas)  
```
