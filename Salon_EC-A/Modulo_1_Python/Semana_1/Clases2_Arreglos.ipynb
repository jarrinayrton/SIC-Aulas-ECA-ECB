## 1. Tuplas (unidad 12, diap. 195-203)
Los datos de la película no deben cambiar, así que se guardan en una **tupla**: parecida a una lista, pero **inmutable**.

```python
pelicula = ("Duna", 155, "13+")  
titulo, duracion, clasificacion = pelicula  
print(titulo, duracion, clasificacion) 
print(pelicula[0])  
print(pelicula[-2:])  

try: 
    pelicula[1] = 160 
except TypeError: 
    print("Error: una tupla no se puede modificar") 

sala_a, sala_b = "Sala 1", "Sala 2"  
sala_a, sala_b = sala_b, sala_a 
print(sala_a, sala_b)  
```

## 2. Lista bidimensional: el mapa de la sala (unidad 13, diap. 224-250)
Una sala es una tabla de filas y columnas. Usamos una **lista de listas**: `0` = asiento libre, `1` = ocupado.

```python
FILAS = 3 
COLUMNAS = 4  
sala = [[0] * COLUMNAS for _ in range(FILAS)]  
sala[0][1] = 1  
sala[2][3] = 1 

for i in range(FILAS):  
    for j in range(COLUMNAS):  
        print(sala[i][j], end=" ")  
    print() 

print(sala[-1]) 
print(sala[0][:2]) 
print(sala[0].count(1))
```

## 3. Listas y sus métodos (unidad 10, diap. 21-55)
Lista de espectadores. Vemos cómo **agregar** (`append`, `extend`), **borrar** (`remove`, `pop`, `del`), **buscar** (`in`), **ordenar** (`sort`) y **rebanar**.

```python
espectadores = ["Carlos", "Ana"] 
espectadores.append("Luis")  
espectadores.extend(["Marta", "Pedro"])  
print(espectadores)  
print(len(espectadores))  
print(espectadores[0], espectadores[-1]) 
print(espectadores[:3]) 

if "Luis" in espectadores:
    espectadores.remove("Luis")
if "Roberto" not in espectadores:
    print("Roberto no está en la lista")

ultimo = espectadores.pop() 
del espectadores[0]  
espectadores.sort()  
print(ultimo, espectadores) 
```

## 4. Diccionarios (unidad 11, diap. 103-117)
Un **diccionario** guarda pares `clave: valor`. Lo usamos para las tarifas y para las reservas. En las reservas, el valor es una **tupla** `(fila, columna, tarifa)`.

```python
precios = {"General": 9, "Estudiante": 5, "Niño": 4} 
print(precios["Estudiante"]) 
precios["Adulto mayor"] = 4 
precios["General"] = 10 
print("Niño" in precios)  
print(10 in precios)  
del precios["Niño"] 
print(len(precios)) 

for tarifa in precios: 
    print("{:<14} ${}".format(tarifa, precios[tarifa]))  

reservas = {"Carlos": (0, 1, "General"), "Ana": (2, 3, "Estudiante")} 
```

## 5. Reserva interactiva
Pedimos datos con `input` y combinamos todo: **ASCII** (`ord`), **rebanado de cadena**, **in / not in**, la **lista 2D**, el **diccionario** y la **lista**. Escribe, por ejemplo, `Julia`, `General` y `B3`.

```python

nombre = input("Nombre: ")  
tarifa = input("Tarifa: ")  
asiento = input("Asiento (ejemplo B3): ")  

fila = ord(asiento[0].upper()) - 65  

columna = int(asiento[1:]) - 1  

if tarifa not in precios:  
    print("Esa tarifa no existe")  
elif fila not in range(FILAS) or columna not in range(COLUMNAS):  
    print("Ese asiento no existe") 
elif sala[fila][columna] == 1:  
    print("Ese asiento ya está ocupado")  
else:  

    sala[fila][columna] = 1  
    reservas[nombre] = (fila, columna, tarifa)  
    espectadores.append(nombre)  
    print("Reserva confirmada. Pago: $", precios[tarifa])
```

## 6. Lista irregular: asientos libres por fila (unidad 13, diap. 239-254)
En una **lista irregular** cada fila puede tener una cantidad distinta de elementos. Aquí cada fila guarda solo los números de butaca libres.

```python
libres_por_fila = []  
for i in range(FILAS):  
    libres = []  
    for j in range(COLUMNAS):  
        if sala[i][j] == 0:  
            libres.append(j + 1)  
    libres_por_fila.append(libres)  
    print("Fila", chr(65 + i), libres)
```

## Resumen
| Estructura | Se escribe con | ¿Se puede modificar? | Se accede con |
|---|---|---|---|
| Lista | `[ ]` | Sí | índice |
| Tupla | `( )` | No | índice |
| Diccionario | `{clave: valor}` | Sí | clave |
| Lista 2D | `[[ ], [ ]]` | Sí | `[fila][columna]` |
