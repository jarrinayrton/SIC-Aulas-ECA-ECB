## 1. Procesar texto con métodos de cadena (unidad 14, diap. 322-338)
Los datos suelen llegar como texto. Con `strip`, `lower`, `split`, `capitalize`, `join`, `startswith` y `replace` los convertimos en algo útil.

```python
registro = "  café:7,bolígrafo:3,leche:1,café:2,pan:5  " 
limpio = registro.strip() 
partes = limpio.split(",")  
print(partes) 

inventario = {} 
for parte in partes:  
    nombre, cantidad = parte.split(":") 
    nombre = nombre.capitalize() 
    inventario.setdefault(nombre, 0) 
    inventario[nombre] += int(cantidad) 
print(inventario) 

print(", ".join(inventario))  
print("Pan".startswith("P"), "Pan".endswith("x")) 
print("1,250,000".replace(",", "."))  
```

## 2. Métodos de diccionario (unidad 14, diap. 296-317)
`update` añade o modifica varios pares, `get` lee sin dar error, `pop` y `popitem` borran devolviendo el valor, y `keys`, `values` e `items` permiten recorrer.

```python
inventario.update({"Pan": 8, "Té": 4}) 
inventario.update(Leche=6)  
print(inventario.get("Sal", 0)) 
quitado = inventario.pop("Té")  
print(quitado, inventario) 
ultimo = inventario.popitem() 
print(ultimo)  

print(list(inventario.keys())) 
print(list(inventario.values()))  
for producto, cantidad in inventario.items(): 
    print("{:<10} {}".format(producto, cantidad))  
```

## 3. defaultdict: contar y agrupar sin comprobar (unidad 15, diap. 356-364)
`defaultdict` crea automáticamente un valor inicial cuando se usa una clave que no existe, así evitamos el `KeyError` y los `if`.

```python

ventas = [("Café", "bebida", 3), ("Pan", "comida", 2), ("Té", "bebida", 4), ("Café", "bebida", 1)] 
total_por_categoria = defaultdict(int)  
productos_por_categoria = defaultdict(list)  

for producto, categoria, unidades in ventas:  
    total_por_categoria[categoria] += unidades  
    productos_por_categoria[categoria].append(producto)  


print(dict(total_por_categoria))  
print(dict(productos_por_categoria))
```

## 4. Diccionarios dobles, copia y deepcopy (unidad 15, diap. 369-376)
Un diccionario puede contener otros diccionarios. Cuidado al copiar: `copy` es superficial, `deepcopy` es completa.

```python
tienda = {"Café": {"precio": 2.5, "cat": "bebida"}, "Pan": {"precio": 1.2, "cat": "comida"}}  
print("Precio original del café:", tienda["Café"]["precio"])  

igual = tienda  
superficial = tienda.copy()  
profunda = copy.deepcopy(tienda)  
tienda["Café"]["precio"] = 99  
print(f"Valor en la variable 'igual': {igual['Café']['precio']} (Refleja el error, es el mismo objeto)")  
print(f"Valor en copia 'superficial': {superficial['Café']['precio']} (Refleja el error, el diccionario interior estaba compartido)")  
print(f"Valor en copia 'profunda': {profunda['Café']['precio']} (Copia segura y 100% independiente)")  
tienda["Café"]["precio"] = 2.5
```

## 7. zip: combinar listas (unidad 16, diap. 456-461)
`zip` junta varias secuencias en tuplas, tomando un elemento de cada una en cada posición.

```python
productos = list(inventario)  
cantidades = list(inventario.values())  
precios = [2.5, 0.8, 1.5, 1.2][:len(productos)] 

total = 0 
for nombre, cant, precio in zip(productos, cantidades, precios): 
    subtotal = cant * precio 
    total += subtotal
    print("{:<10} {:>3} x {:>4} = {:>6.2f}".format(nombre, cant, precio, subtotal))  

print("Total del inventario:", round(total, 2))  
catalogo = dict(zip(productos, precios))  
print(catalogo)  

pares = list(zip(productos, cantidades)) 
nombres, cants = zip(*pares)
print(nombres, cants)  
```
