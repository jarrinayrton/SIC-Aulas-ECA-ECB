# EJERCICIO 1 · Factorial y Fibonacci


### 1.1 Nuestra primera función: Fibonacci con bucle

entender cómo se define, se llama y se devuelve un resultado en una función, y tener la versión con bucle como referencia de velocidad.

```python
def fibonacci_bucle(n=10):  
    anterior, actual = 0, 1  
    for _ in range(n):  
        anterior, actual = actual, anterior + actual  
    return anterior 

print(fibonacci_bucle(10))  
print(fibonacci_bucle())  
print(fibonacci_bucle(n=7))  

N = 30  
inicio = time.perf_counter() 
resultado_bucle = fibonacci_bucle(N) 
t_bucle = time.perf_counter() - inicio  
print("Bucle    F({}) = {} | tiempo: {:.8f} s".format(N, resultado_bucle, t_bucle)) 
```

### 1.2 Funciones que usan otras funciones y devuelven varios valores

ver que una función puede llamar a otra, que `return` puede devolver varios valores en una tupla y cómo se reciben.

```python
def terminos_y_suma(cantidad):  
    terminos = [fibonacci_bucle(i) for i in range(cantidad)] 
    return terminos, sum(terminos)  

lista, suma = terminos_y_suma(10)  
print(lista)  
print("Suma:", suma)  
```

### 1.3 Factorial con bucle

tener la versión con bucle para compararla con la versión recursiva 

```python
def factorial_bucle(n): 
    resultado = 1  
    for numero in range(2, n + 1):  
        resultado *= numero 
    return resultado

print(factorial_bucle(5)) 
print(factorial_bucle(0))  
```

### 1.4 Factorial con recursión

entender el caso base, la llamada recursiva y cómo los resultados vuelven hacia arriba 

```python
def factorial_recursivo(n): 
    if n <= 1:  # CASO BASE
        return 1 
    return n * factorial_recursivo(n - 1) 

print(factorial_recursivo(5))  
print(factorial_recursivo(5) == factorial_bucle(5)) 
```

### 1.5 Fibonacci con recursión (y su tiempo)

comprobar con números reales que la recursión simple es **mucho más lenta** que el bucle (diap. 114).

```python
def fibonacci_recursivo(n): 
    if n <= 1:  # CASOS BASE
        return n  
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)  

inicio = time.perf_counter()  
resultado_rec = fibonacci_recursivo(N) 
t_rec = time.perf_counter() - inicio  
print("Recursivo F({}) = {} | tiempo: {:.8f} s".format(N, resultado_rec, t_rec)) 
print("¿Mismo resultado que el bucle?", resultado_rec == resultado_bucle) 
print("El recursivo tardó unas {:.0f} veces más que el bucle".format(t_rec / t_bucle)) 
```

### 1.6 Fibonacci recursivo con memorización (y su tiempo)

 ver que una pequeña idea (recordar resultados) vuelve a la recursión casi tan rápida como el bucle.

```python
memoria = {}  

def fibonacci_memo(n):  
    if n <= 1:  # casos base
        return n 
    if n not in memoria:  
        memoria[n] = fibonacci_memo(n - 1) + fibonacci_memo(n - 2)  
    return memoria[n]  

memoria.clear()  
inicio = time.perf_counter()  
resultado_memo = fibonacci_memo(N) 
t_memo = time.perf_counter() - inicio  
inicio = time.perf_counter() 
fibonacci_memo(N)  
t_memo2 = time.perf_counter() - inicio  
print("Memoria  F({}) = {} | tiempo: {:.8f} s".format(N, resultado_memo, t_memo))  
print("Segunda llamada (ya en memoria): {:.8f} s".format(t_memo2)) 
print("Valores guardados en memoria:", len(memoria))  
```

concluir cuándo conviene cada técnica: bucle, recursión simple o recursión con memorización (diap. 109).

# EJERCICIO 2 · Libreta de notas (sencillo)


### 2.1 Variables globales y locales



```python
NOTA_MINIMA = 60 
evaluados = 0  

def evaluar(nota): 
    global evaluados 
    evaluados += 1 
    estado = "aprobado" if nota >= NOTA_MINIMA else "reprobado" 
    return estado  

print(evaluar(80), evaluar(40), "| evaluados:", evaluados)  
```

### 2.2 lambda, filter, map, reduce y comprensión de listas


```python
notas = [85, 42, 70, 95, 58, 60] 

aprobadas = list(filter(lambda n: n >= 60, notas))  
con_bonus = list(map(lambda n: min(n + 5, 100), notas))  
total = reduce(lambda acumulado, n: acumulado + n, notas) 
print(aprobadas, con_bonus, total) 

print([n for n in notas if n >= 60] == aprobadas)  
print([min(n + 5, 100) for n in notas] == con_bonus)  
print(sorted(notas, reverse=True), all(n > 0 for n in notas), any(n < 50 for n in notas))  
```
